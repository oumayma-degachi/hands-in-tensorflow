# Using_fashion_mnist_dataset
In this project, we build a simple neural network to differenciate 10 different clothing items.
We are using fashion_mnist_dataset. each element is identified by 28x28 graycell pixels.


The [Fashion MNIST dataset](https://github.com/zalandoresearch/fashion-mnist) is a collection of grayscale 28x28 pixel clothing images. Each image is associated with a label as shown in this table.

| Label | Description |
| --- | --- |
| 0 | T-shirt/top |
| 1 | Trouser |
| 2 | Pullover |
| 3 | Dress |
| 4 | Coat |
| 5 | Sandal |
| 6 | Shirt |
| 7 | Sneaker |
| 8 | Bag |
| 9 | Ankle boot |

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
```

# Load the Fashion MNIST dataset

```python
fmnist = tf.keras.datasets.fashion_mnist
```

# Load the training and test split of the Fashion MNIST dataset

```python
(training_images, training_labels), (test_images, test_labels) = fmnist.load_data()
```

# Print a training image and a training label to see.


```python
# You can put between 0 to 59999 here
index = 0

# Set number of characters per row when printing
np.set_printoptions(linewidth=320)

# Print the label and image
print(f'LABEL: {training_labels[index]}')
print(f'\nIMAGE PIXEL ARRAY:\n\n{training_images[index]}\n\n')

# Visualize the image using the default colormap (viridis)
plt.imshow(training_images[index])
plt.colorbar()
plt.show()
```

# Normalization

```python
# Normalize the pixel values of the train and test images
training_images  = training_images / 255.0
test_images = test_images / 255.0
```

# Build the model

```python
# Build the classification model
model = tf.keras.models.Sequential([
    tf.keras.Input(shape=(28,28)),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(128, activation=tf.nn.relu),
    tf.keras.layers.Dense(10, activation=tf.nn.softmax)
])
```

# Compile and Train the model

```python
model.compile(optimizer = tf.optimizers.Adam(),
              loss = 'sparse_categorical_crossentropy',
              metrics=['accuracy'])

model.fit(training_images, training_labels, epochs=5)
```

# Evaluate the model on unseen data

```python
model.evaluate(test_images, test_labels)
```
