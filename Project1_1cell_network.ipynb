## Project1_1cell_network.ipynb

Let’s write code for a function like this:

def f_function(x):
    y = (3 * x) - 2
    return y

We are going to feed data to our network so that it can learn this task.
We provide a set of x and y values, and after training, the network will learn the relationship between x and y.


## Imports

Start with the imports.
- TensorFlow called `tf` for convention and ease of use (keras is contained within tensorflow)
- numpy to represent data as arrays easily and to optimize numerical operations.

```python
import tensorflow as tf
import numpy as np
```

## Generating the Data


```python
# Declare model inputs and outputs for training
xs = np.array([-1.0,  0.0, 1.0, 2.0, 3.0, 4.0], dtype=float)
ys = np.array([-5.0, -2.0, 1.0, 4.0, 7.0, 10.0], dtype=float)
```

## Define and Compile the Neural Network

Next, you will create the simplest possible neural network. It has 1 layer with 1 neuron.


```python
# Build a simple Sequential model
model = tf.keras.Sequential([

    # Define the input shape
    tf.keras.Input(shape=(1,)),

    # Add a Dense layer
    tf.keras.layers.Dense(units=1)
    ])
```

Now, you will compile the neural network. When you do so, you have to specify 2 functions: a loss and an optimizer



```python
# Compile the model
model.compile(optimizer='sgd', loss='mean_squared_error')
```

# Training the Neural Network

The process of training the neural network, where it learns the relationship between the x's and y's is in the model.fit().

```python
# Train the model
model.fit(xs, ys, epochs=400)
```

Ok, now you have a model that has been trained to learn the relationship between `x` and `y`. We can also make a prediction to an unknon value. Let's check what it finds for x=10 (y should be 27)

```python
# Make a prediction
print(f"model predicted: {model.predict(np.array([10.0]), verbose=2).item():.5f}")
```
